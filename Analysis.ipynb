```python
# ============================================
# AUTO MPG DATASET ANALYSIS (FULL SINGLE CODE)
# ============================================

# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load Dataset (YOUR PATH)
df = pd.read_csv(r"D:\\Project\\TASK_22_SAYAYN\\auto-mpg.csv")

print("First 5 Rows:")
print(df.head())

print("\nDataset Info:")
print(df.info())

# 3. Fix Column Names (remove spaces)
df.columns = df.columns.str.replace(" ", "_")

# 4. Data Cleaning
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
df = df.dropna()

# Convert origin numbers to country names
df['origin'] = df['origin'].replace({1:'USA', 2:'Europe', 3:'Japan'})

print("\nDataset Shape After Cleaning:", df.shape)

# ============================================
# 5. KPI CALCULATIONS
# ============================================

total_cars = df.shape[0]
avg_mpg = df['mpg'].mean()
avg_hp = df['horsepower'].mean()

print("\n===== KPI RESULTS =====")
print("Total Vehicles:", total_cars)
print("Average MPG:", round(avg_mpg,2))
print("Average Horsepower:", round(avg_hp,2))

# MPG by Year
mpg_year = df.groupby('model_year')['mpg'].mean()

# ============================================
# 6. Fuel Efficiency Trend
# ============================================

plt.figure()
plt.plot(mpg_year)
plt.title("Fuel Efficiency Trend")
plt.xlabel("Model Year")
plt.ylabel("Average MPG")
plt.show()

# ============================================
# 7. MPG by Origin
# ============================================

plt.figure()
df.groupby('origin')['mpg'].mean().plot(kind='bar')
plt.title("Average MPG by Origin")
plt.xlabel("Origin")
plt.ylabel("MPG")
plt.show()

# ============================================
# 8. Market Share by Origin
# ============================================

plt.figure()
df['origin'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Market Share by Origin")
plt.ylabel("")
plt.show()

# ============================================
# 9. MPG vs Weight
# ============================================

plt.figure()
sns.scatterplot(data=df, x='weight', y='mpg')
plt.title("MPG vs Weight")
plt.show()

# ============================================
# 10. MPG vs Horsepower
# ============================================

plt.figure()
sns.scatterplot(data=df, x='horsepower', y='mpg')
plt.title("MPG vs Horsepower")
plt.show()

# ============================================
# 11. Engine Distribution
# ============================================

plt.figure()
df['cylinders'].value_counts().plot(kind='bar')
plt.title("Engine Distribution")
plt.xlabel("Cylinders")
plt.ylabel("Vehicle Count")
plt.show()

# ============================================
# 12. MPG Distribution
# ============================================

plt.figure()
df['mpg'].plot(kind='hist', bins=20)
plt.title("MPG Distribution")
plt.xlabel("MPG")
plt.show()

# ============================================
# 13. Correlation Heatmap
# ============================================

corr = df[['mpg','weight','horsepower','displacement','cylinders','acceleration']].corr()

plt.figure()
sns.heatmap(corr, annot=True)
plt.title("Correlation Matrix")
plt.show()

# ============================================
# 14. Insights
# ============================================

print("\n===== INSIGHTS =====")
print("1. Fuel efficiency improves across model years.")
print("2. Japanese cars generally have higher MPG.")
print("3. Higher vehicle weight reduces fuel efficiency.")
print("4. Higher horsepower leads to lower MPG.")
print("5. 4-cylinder engines dominate the dataset.")
```
